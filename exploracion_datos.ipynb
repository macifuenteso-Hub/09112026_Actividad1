```python
import pandas as pd

print("¡Python está funcionando correctamente!") 
```

```python
# Dirección del archivo CSV
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"

# Cargar el conjunto de datos
datos = pd.read_csv(url)

print("¡Datos cargados correctamente!")
```

```python
# Mostrar las primeras cinco filas
datos.head()
```

```python
# Conocer el número de filas y columnas
print("Dimensiones:", datos.shape)

# Mostrar los nombres de las columnas
print("Columnas:", datos.columns.tolist())
```

```python
#### Descripción del conjunto de datos Se utilizó el conjunto de datos *Palmer Penguins*, que contiene información sobre tres especies de pingüinos (Adelie, Chinstrap y Gentoo) observadas en el archipiélago Palmer, en la Antártida.

#El conjunto de datos contiene 344 registros y 7 variables, entre ellas:

#**species:** especie de pingüino
#**island:** isla donde se registró
#**bill_length_mm:** longitud del pico en milímetros
#**bill_depth_mm:** profundidad del pico en milímetros
#**flipper_length_mm:** longitud de la aleta en milímetros
#**body_mass_g:** masa corporal en gramos
#**sex:** sexo del individuo

#Este conjunto de datos permite explorar información biológica y realizar comparaciones entre especies mediante herramientas de análisis de datos en Python.

#**Fuente:** Palmer Penguins, datos disponibles a través del repositorio público `seaborn-data` en GitHub.
```

```python
# Guardar los datos en un archivo CSV
datos.to_csv("pinguinos.csv", index=False)

print("Archivo pinguinos.csv guardado correctamente")
```
